```python
# suprevised learning
rainfull = 15
river_basin = 5
total_water = rainfull + river_basin
print(total_water)
# Step 1: Import the machine learning tools
import numpy as np
from sklearn.linear_model import LinearRegression

# Step 2: Provide historical data (Rainfall vs River Flow)
# Rainfall data (Input x)
X = np.array([[1], [2], [3], [4], [5]])
# River Flow data (Output y)
y = np.array([5, 8, 11, 14, 17])

# Step 3: Create the Linear Regression Machine
model = LinearRegression()

# Step 4: Train the machine to find the formula (Fit the model)
model.fit(X, y)

# Step 5: Print the hidden numbers discovered by the machine (w and b)
print("Discovered Weight (w):", model.coef_[0])
print("Discovered Bias (b):", model.intercept_)

# Step 6: Predict river flow for a new rainfall of 6 mm
new_rainfall = np.array([[6]])
prediction = model.predict(new_rainfall)
print("Predicted River Flow for 6mm rainfall:", prediction[0])
```

```python
# Step 1: Import the error measurement tool
import numpy as np
from sklearn.metrics import mean_squared_error

# Step 2: Define actual river flow vs machine's predicted flow
actual_flow = np.array([10, 15, 20])
predicted_flow = np.array([8, 16, 17])

# Step 3: Calculate the Mean Squared Error (MSE)
mse_result = mean_squared_error(actual_flow, predicted_flow)

# Step 4: Print the error score
print("Machine Error Score (MSE):", mse_result)
```

```python
# Step 1: Import the Stochastic Gradient Descent tool
import numpy as np
from sklearn.linear_model import SGDRegressor

# Step 2: Define historical irrigation data
# Water input speed (X) and Canal Depth (y)
X = np.array([[10], [20], [30], [40]])
y = np.array([5, 9, 13, 17])

# Step 3: Create the machine with a specific learning rate (alpha = 0.01)
# max_iter is the number of steps (epochs) the machine takes down the valley
model = SGDRegressor(max_iter=5000, eta0=0.0001)
# Step 4: Train the model using Gradient Descent
model.fit(X, y)

# Step 5: Check what the machine learned after the steps
print("Final Weight after Gradient Descent (w):", model.coef_)
print("Final Intercept after Gradient Descent (b):", model.intercept_)
```

```python
# Step 1: Import the Logistic Regression classification tool
import numpy as np
from sklearn.linear_model import LogisticRegression

# Step 2: Provide historical Soil Moisture data (Input X) and Drought labels (Output y)
# X: Soil Moisture % (10%, 15% are dry | 35%, 40% are wet)
X = np.array([[10], [15], [35], [40]])

# y: 1 means Drought (Yes), 0 means No Drought (No)
y = np.array([1, 1, 0, 0])

# Step 3: Create the Logistic Regression Machine (contains Sigmoid inside)
clf_model = LogisticRegression()

# Step 4: Train the machine
clf_model.fit(X, y)

# Step 5: Predict for a new farm with only 12% Soil Moisture
new_farm = np.array([[12]])
prediction = clf_model.predict(new_farm)

# Step 6: Check the exact probability score behind the prediction
prob_score = clf_model.predict_proba(new_farm)

print("Is there a Drought? (1=Yes, 0=No):", prediction)
print("Probability of [No Drought, Yes Drought]:", prob_score)
```

```python
import numpy as np
import torch

# Define 3 separate satellite bands as individual clean values
band_red = 0.15
band_nir = 0.45
band_swir = 0.10

# Define 3 hidden weights discovered by the neural network
weight_red = 2.0
weight_nir = 5.0
weight_swir = -1.0

bias_factor = 0.50

# Pack safely into flat input arrays
inputs_x = np.array([band_red, band_nir, band_swir])
weights_w = np.array([weight_red, weight_nir, weight_swir])

# 1. Calculate Dot Product using NumPy
numpy_z = np.dot(inputs_x, weights_w) + bias_factor

# 2. Calculate Dot Product using PyTorch Tensors
torch_x = torch.tensor([band_red, band_nir, band_swir])
torch_w = torch.tensor([weight_red, weight_nir, weight_swir])
pytorch_z = torch.dot(torch_x, torch_w) + bias_factor

print("NumPy Dot Product Engine (z):", numpy_z)
print("PyTorch Dot Product Engine (z):", pytorch_z.item())
```

```python
import numpy as np
import torch
import tensorflow as tf
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_squared_error, classification_report
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

print("========================================================================================")
print("                    ZTerraIQ GeoAI & HYDROINFORMATICS OPERATIONAL SYSTEMS               ")
print("========================================================================================\n")

# ----------------------------------------------------------------------------------------
# [SYSTEM 1] -> RIVER BASIN STREAMFLOW PREDICTOR
# ----------------------------------------------------------------------------------------
print("[RUNNING SYSTEM 1: RIVER BASIN STREAMFLOW PREDICTOR...]")

# Inputs: Clean, flat arrays for 3 upstream stations across 4 separate observation days
day1_stations = [12.5, 8.0, 4.2]
day2_stations = [22.1, 14.5, 9.0]
day3_stations = [5.0, 2.1, 1.0]
day4_stations = [34.8, 20.2, 15.6]

station_inputs = np.array([day1_stations, day2_stations, day3_stations, day4_stations])
actual_flow = np.array([125.4, 210.8, 45.2, 315.6])

reg_model = LinearRegression()
reg_model.fit(station_inputs, actual_flow)
flow_predictions = reg_model.predict(station_inputs)

project1_mse = mean_squared_error(actual_flow, flow_predictions)
print(f"-> Discovered Station Weights (w): {reg_model.coef_}")
print(f"-> Verification Metric - Mean Squared Error (MSE): {project1_mse:.4f}")
print("-> SYSTEM 1 STATUS: VERIFIED OPERATION\n")


# ----------------------------------------------------------------------------------------
# [SYSTEM 2] -> AWASH VALLEY DROUGHT CLASSIFIER
# ----------------------------------------------------------------------------------------
print("[RUNNING SYSTEM 2: AWASH VALLEY DROUGHT CLASSIFIER...]")

# Inputs: Clean, flat arrays for [Soil Moisture %, Temperature C] over 5 regional grid locations
loc1_features = [10.5, 38.5]
loc2_features = [35.2, 24.1]
loc3_features = [12.1, 39.2]
loc4_features = [40.8, 22.5]
loc5_features = [15.4, 36.0]

environment_features = np.array([loc1_features, loc2_features, loc3_features, loc4_features, loc5_features])
drought_labels = np.array([1, 0, 1, 0, 1])

clf_model = LogisticRegression(random_state=42)
clf_model.fit(environment_features, drought_labels)
drought_predictions = clf_model.predict(environment_features)

print("-> Verification Metric - Classification Quality Report:")
print(classification_report(drought_labels, drought_predictions))
print("-> SYSTEM 2 STATUS: VERIFIED OPERATION\n")


# ----------------------------------------------------------------------------------------
# [SYSTEM 3] -> SATELLITE LAND COVER SEGMENTATION MAPPER
# ----------------------------------------------------------------------------------------
print("[RUNNING SYSTEM 3: SATELLITE LAND COVER SEGMENTATION MAPPER...]")

# Inputs: Clean, flat arrays for Sentinel-2 bands [Red_Band, NIR_Band] for 4 unclassified pixels
pixel1_bands = [0.12, 0.15]
pixel2_bands = [0.10, 0.14]
pixel3_bands = [0.08, 0.55]
pixel4_bands = [0.09, 0.62]

sentinel_pixels = np.array([pixel1_bands, pixel2_bands, pixel3_bands, pixel4_bands])

cluster_engine = KMeans(n_clusters=2, random_state=42)
discovered_groups = cluster_engine.fit_predict(sentinel_pixels)

print(f"-> Segmentation Output Labels: {discovered_groups}")
print("-> SYSTEM 3 STATUS: VERIFIED OPERATION\n")


# ----------------------------------------------------------------------------------------
# [SYSTEM 4] -> HYPERSPECTRAL DIMENSIONALITY COMPRESSION PIPELINE
# ----------------------------------------------------------------------------------------
print("[RUNNING SYSTEM 4: HYPERSPECTRAL DIMENSIONALITY COMPRESSION PIPELINE...]")

# Inputs: Clean, flat arrays for 4 separate channels containing dense hyperspectral sensor matrices
chan1_matrix = [50, 50, 50, 50]
chan2_matrix = [205, 205, 205, 205]
chan3_matrix = [35, 35, 35, 35]
chan4_matrix = [175, 175, 175, 175]

hyperspectral_matrix = np.array([chan1_matrix, chan2_matrix, chan3_matrix, chan4_matrix])

pca_pipeline = PCA(n_components=2, random_state=42)
compressed_matrix = pca_pipeline.fit_transform(hyperspectral_matrix)

print(f"-> Input Data Channels: {hyperspectral_matrix.shape}")
print(f"-> Compressed Operational Channels: {compressed_matrix.shape}")
print("-> SYSTEM 4 STATUS: VERIFIED OPERATION\n")

print("========================================================================================")
print("             🎉 ALL SYSTEMS RUNNING AND VERIFIED SUCCESSFULLY! 🎉              ")
print("========================================================================================")
```
